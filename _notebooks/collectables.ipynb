```python
import java.util.ArrayList;
import java.util.HashMap;
import java.util.HashSet;
import java.util.List;
import java.util.Map;
import java.util.Set;
import java.util.Stack;

public class Collectible {
    private String name;
    private String type;
    private int rarity;
    private double value;

    public Collectible(String name, String type, int rarity, double value) {
        this.name = name;
        this.type = type;
        this.rarity = rarity;
        this.value = value;
    }

    public String getName() { return name; }
    public String getType() { return type; }
    public int getRarity() { return rarity; }
    public double getValue() { return value; }
    
    public void setValue(double value) { this.value = value; }

    public String toString() {
        return String.format("%-15s [%-6s] (Rarity: %d, Value: $%.2f)", 
                            name, type, rarity, value);
    }
}

public class Inventory {
    private ArrayList<Collectible> items;
    private Set<String> tradedItems;
    private Set<String> duplicateItems;
    private Map<String, Collectible> itemDetails;
    private Stack<String> folderStack;

    public Inventory() {
        items = new ArrayList<>();
        tradedItems = new HashSet<>();
        duplicateItems = new HashSet<>();
        itemDetails = new HashMap<>();
        folderStack = new Stack<>();
        folderStack.push("Root");
    }

    // Part 1: List Task - Bulk add and remove
    public void addAllItems(List<Collectible> newItems) {
        for (Collectible item : newItems) {
            addItem(item);
        }
    }

    public void removeAllItems(List<Collectible> itemsToRemove) {
        items.removeAll(itemsToRemove);
    }

    // Part 2: Dictionary Task - Value updates and rarity filters
    public void updateValuesByRarity(int rarity, double percentage) {
        for (Collectible item : items) {
            if (item.getRarity() == rarity) {
                double newValue = item.getValue() * (1 + percentage/100);
                item.setValue(newValue);
                itemDetails.put(item.getName(), item);
            }
        }
    }

    public List<Collectible> filterByRarity(int rarity) {
        List<Collectible> result = new ArrayList<>();
        for (Collectible item : items) {
            if (item.getRarity() == rarity) {
                result.add(item);
            }
        }
        return result;
    }

    // Part 3: Set Task - Track traded and duplicate items
    public void addItem(Collectible c) {
        if (items.contains(c)) {
            duplicateItems.add(c.getName());
            System.out.println("Duplicate detected: " + c.getName());
            return;
        }
        items.add(c);
        itemDetails.put(c.getName(), c);
    }

    public void markAsTraded(String itemName) {
        tradedItems.add(itemName);
    }

    // Part 4: Print Method - Formatted display
    public void displayCollection() {
        System.out.println("Current Folder: " + getCurrentPath());
        System.out.println("--------------------------------------------------");
        System.out.printf("%-15s %-10s %-10s %-10s%n", 
                         "Name", "Type", "Rarity", "Value");
        System.out.println("--------------------------------------------------");
        for (Collectible item : items) {
            System.out.printf("%-15s %-10s %-10d $%-10.2f%n",
                             item.getName(), item.getType(), 
                             item.getRarity(), item.getValue());
        }
        System.out.println("--------------------------------------------------");
        System.out.println("Total items: " + items.size());
        System.out.println("Traded items: " + tradedItems.size());
        System.out.println("Duplicate items detected: " + duplicateItems.size());
    }

    // Bonus Challenge: Stack-based folder navigation
    public void enterFolder(String folderName) {
        folderStack.push(folderName);
    }

    public void exitFolder() {
        if (folderStack.size() > 1) {
            folderStack.pop();
        }
    }

    public String getCurrentPath() {
        return String.join(" > ", folderStack);
    }
}

public class Main {
    public static void main(String[] args) {
        // Create inventory
        Inventory inventory = new Inventory();
        
        // Create some collectibles
        Collectible item1 = new Collectible("Phoenix Flame", "Fire", 5, 100.0);
        Collectible item2 = new Collectible("Aqua Sprite", "Water", 3, 50.0);
        Collectible item3 = new Collectible("Blazing Ember", "Fire", 2, 30.0);
        Collectible item4 = new Collectible("Dragon Scale", "Fire", 5, 150.0);
        Collectible item5 = new Collectible("Phoenix Flame", "Fire", 5, 100.0); // Duplicate
        
        // Add items individually
        inventory.addItem(item1);
        inventory.addItem(item2);
        inventory.addItem(item3);
        
        // Add items in bulk (List Task)
        List<Collectible> bulkItems = new ArrayList<>();
        bulkItems.add(item4);
        bulkItems.add(item5);
        inventory.addAllItems(bulkItems);
        
        // Mark some items as traded
        inventory.markAsTraded("Aqua Sprite");
        inventory.markAsTraded("Dragon Scale");
        
        // Update values by rarity (Dictionary Task)
        inventory.updateValuesByRarity(5, 10); // Increase legendary items by 10%
        
        // Display the collection
        inventory.displayCollection();
        
        // Filter by rarity (Dictionary Task)
        System.out.println("\nLegendary items (Rarity 5):");
        List<Collectible> legendaries = inventory.filterByRarity(5);
        for (Collectible item : legendaries) {
            System.out.println(item);
        }
        
        // Folder navigation (Bonus Challenge)
        System.out.println("\nFolder Navigation:");
        inventory.enterFolder("Weapons");
        inventory.enterFolder("Magic");
        inventory.enterFolder("Fire");
        System.out.println("Current path: " + inventory.getCurrentPath());
        
        inventory.exitFolder();
        System.out.println("After exiting: " + inventory.getCurrentPath());
    }
}

Main.main(null);
```
